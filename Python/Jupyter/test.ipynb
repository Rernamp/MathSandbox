```python
import matplotlib.pyplot as plt
import numpy as np
```

# Init varaible

```python
frequency = 1
sampla_rate = 100
amplitude = 1
duration_s = 2
points_count = int(sampla_rate * duration_s)
time = np.linspace(0, duration_s, points_count)

y = amplitude * np.sin(2 * np.pi * frequency * time)

```

## Plot graph

```python
plt.plot(time, y)
plt.grid(True)
```
